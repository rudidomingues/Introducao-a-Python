# Aula: Introdução a Python

## Contextualização

Python é uma linguagem de programação versátil e fácil de aprender. É amplamente utilizada em ciência de dados, automação, desenvolvimento web, robótica e muito mais. Nesta aula, vamos explorar seus principais conceitos e fazer atividades práticas para aplicar o que foi visto.

## Exemplo prático

Execute a célula abaixo para ver um exemplo simples de código Python. O código calcula a soma de dois números e imprime o resultado.
