## 1. O que é programar?

Programar é criar uma sequência de instruções para que o computador execute uma tarefa.

Exemplo do cotidiano:

**Problema:** calcular a média de um estudante.  
**Entrada:** duas notas.  
**Processamento:** somar as notas e dividir por 2.  
**Saída:** mostrar a média final.

Esse raciocínio é conhecido como **pensamento computacional**, pois envolve decompor o problema, identificar dados importantes, criar passos lógicos e testar a solução.

## 4. Entrada de dados

O comando `input()` permite que o usuário digite uma informação.

## 5. Operadores matemáticos

| Operador | Função | Exemplo |
|---|---|---|
| `+` | soma | `2 + 3` |
| `-` | subtração | `5 - 2` |
| `*` | multiplicação | `4 * 3` |
| `/` | divisão | `10 / 2` |
| `//` | divisão inteira | `10 // 3` |
| `%` | resto da divisão | `10 % 3` |
| `**` | potência | `2 ** 3` |

````python
a = 10
b = 3

print("Soma:", a + b)
print("Subtração:", a - b)
print("Multiplicação:", a * b)
print("Divisão:", a / b)
print("Divisão inteira:", a // b)
print("Resto da divisão:", a % b)
print("Potência:", a ** b)
````

## 6. Exemplo prático: cálculo de média

Agora vamos criar um pequeno programa para calcular a média de um estudante.

````python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

print("A média do estudante é:", media)
````

## 7. Estrutura condicional: tomada de decisão

A estrutura `if`, `elif` e `else` permite que o programa tome decisões.

Exemplo:

- se a média for maior ou igual a 7, o estudante está aprovado;
- se a média for maior ou igual a 5, está em recuperação;
- caso contrário, está reprovado.

````python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

print("Média:", media)

if media >= 7:
    print("Situação: Aprovado")
elif media >= 5:
    print("Situação: Recuperação")
else:
    print("Situação: Reprovado")
````

## 8. Teste de mesa

O teste de mesa é uma técnica usada para acompanhar manualmente o comportamento de um algoritmo.

Ele ajuda a verificar:

- quais valores entram no programa;
- como as variáveis mudam;
- qual será a saída;
- se a lógica está correta.

### Exemplo de algoritmo

```python
nota1 = 8
nota2 = 6
media = (nota1 + nota2) / 2

if media >= 7:
    print("Aprovado")
else:
    print("Reprovado")
```

### Teste de mesa

| Passo | Variável `nota1` | Variável `nota2` | Variável `media` | Condição | Saída |
|---|---:|---:|---:|---|---|
| 1 | 8 | - | - | - | - |
| 2 | 8 | 6 | - | - | - |
| 3 | 8 | 6 | 7 | `media >= 7` é verdadeiro | Aprovado |
